```python
violence_visual_data = np.load('data/images/xd_violence/violence_visual_data.npy')
violence_audio_data = np.load('data/images/xd_violence/violence_audio_data.npy')
normal_visual_data = np.load('data/images/xd_violence/normal_visual_data.npy')
normal_audio_data = np.load('data/images/xd_violence/normal_audio_data.npy')
```

```python
from sklearn.model_selection import train_test_split
from training.kfold_train import stratified_kfold_train_val

from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier

from sklearn.metrics import confusion_matrix, classification_report
```

```python
violence_labels = np.ones(len(violence_visual_data))
normal_labels = np.zeros(len(normal_visual_data))

violence_visual_data = np.array(violence_visual_data)
normal_visual_data = np.array(normal_visual_data)
```

```python
x = np.concat([violence_visual_data, normal_visual_data])
y = np.concat([violence_labels, normal_labels])
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y,
                                                    test_size=0.2)
```

```python
import mlflow
import mlflow.sklearn
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report, 
    confusion_matrix, 
    accuracy_score, 
    precision_score, 
    recall_score, 
    f1_score
)

def stratified_kfold_train_val_mlflow(folds, threshold, model_params, x_train, y_train):
    """
    Executes Stratified K-Fold CV and logs every fold as a nested child run in MLflow.
    Returns the mean aggregate metrics across all folds.
    """
    # Initialize the Scikit-Learn splitter
    skf = StratifiedKFold(n_splits=folds, shuffle=True, random_state=model_params["random_state"])
    
    # Storage arrays for summary calculations
    cv_precision = []
    cv_recall = []

    print(f"🔄 Starting {folds}-Fold Cross Validation Tracking...")

    for fold_idx, (train_idx, val_idx) in enumerate(skf.split(x_train, y_train)):
        # 🔥 THE SECRET: Start a nested child run dedicated to this single fold
        with mlflow.start_run(run_name=f"Fold_{fold_idx + 1}", nested=True):
            
            # Split the training matrix into fold splits
            X_tr, X_val = x_train[train_idx], x_train[val_idx]
            y_tr, y_val = y_train[train_idx], y_train[val_idx]
            
            # Re-instantiate a fresh clone of the model for this fold
            fold_model = LogisticRegression(
                penalty=model_params["penalty"],
                C=model_params["C"],
                solver=model_params["solver"],
                max_iter=model_params["max_iter"],
                random_state=model_params["random_state"],
                class_weight=model_params["class_weight"]
            )
            
            # Train model on fold subset
            fold_model.fit(X_tr, y_tr)
            
            # Evaluate on verification fold split
            probs = fold_model.predict_proba(X_val)[:, 1]
            preds = (probs >= threshold).astype(int)
            
            # Calculate metrics for this specific fold validation block
            fold_precision = precision_score(y_val, preds, average='binary')
            fold_recall = recall_score(y_val, preds, average='binary')
            
            # Log metrics straight to the child run panel
            mlflow.log_metric("fold_val_precision", fold_precision)
            mlflow.log_metric("fold_val_recall", fold_recall)
            
            # Keep historical arrays updated
            cv_precision.append(fold_precision)
            cv_recall.append(fold_recall)
            
            print(f"   ↳ Fold {fold_idx + 1} Complete | Val Precision: {fold_precision:.4f} | Val Recall: {fold_recall:.4f}")
            
    # Return aggregated cross-validation metrics back to the parent execution line
    return np.mean(cv_precision), np.mean(cv_recall)
```

```python
# Initialize or connect to a local lightweight SQLite database fill
mlflow.set_tracking_uri("sqlite:////Users/ngurahbagus/Python/Stage 1/mlflow.db")

# 2. Declare your experiment name
mlflow.set_experiment("Multimodal Experiment 2: Logistic Regression Classifier for XD Violence Video Array")

# 2. Master Parameter Configuration Bundle
params = {
    "penalty": "l2",
    "C": 0.5,
    "solver": "liblinear",
    "max_iter": 1000,
    "random_state": 42,
    "class_weight": "balanced",
    "classification_threshold": 0.30,
    "k_folds": 5
}

# 3. Open the Parent Run Environment
with mlflow.start_run(run_name="logistic_regression_with_cv_tree"):
    
    # Log your configurations at the parent level
    mlflow.log_params(params)
    
    # ─── STEP A: RUN NESTED CROSS-VALIDATION ───
    # This runs the loop and writes child runs directly under this active context block
    mean_cv_precision, mean_cv_recall = stratified_kfold_train_val_mlflow(
        folds=params["k_folds"],
        threshold=params["classification_threshold"],
        model_params=params,
        x_train=x_train,
        y_train=y_train
    )
    
    # Save the cross-validation aggregate averages to your primary parent sheet tracking matrix
    mlflow.log_metric("cv_mean_precision", mean_cv_precision)
    mlflow.log_metric("cv_mean_recall", mean_cv_recall)

    # ─── STEP B: TRAIN SYSTEM-WIDE PRODUCTION CHECKPOINT ───
    print("\n⏳ Training production model on entire train split...")
    final_model = LogisticRegression(
        penalty=params["penalty"],
        C=params["C"],
        solver=params["solver"],
        max_iter=params["max_iter"],
        random_state=params["random_state"],
        class_weight=params["class_weight"]
    )
    final_model.fit(x_train, y_train)
    
    # Evaluate final hold-out test metrics
    test_probabilities = final_model.predict_proba(x_test)[:, 1]
    test_predictions = (test_probabilities >= params["classification_threshold"]).astype(int)

    # Calculate final performance metrics
    test_precision = precision_score(y_test, test_predictions, average='binary')
    test_recall = recall_score(y_test, test_predictions, average='binary')

    # Log clean columns to the main dashboard row entry
    mlflow.log_metric("final_test_precision", test_precision)
    mlflow.log_metric("final_test_recall", test_recall)

    # ─── STEP C: SAVE FINAL ARTIFACT CHECKPOINT ───
    mlflow.sklearn.log_model(final_model, artifact_path="final_production_head")

    # Notebook standard validation diagnostic blocks
    print("\n📊 Final Production Hold-Out Classification Report:")
    print(classification_report(y_test, test_predictions))
    print("📊 Confusion Matrix:")
    print(confusion_matrix(y_test, test_predictions))
    print(f"\n🚀 Complete cross-validation structure tree saved perfectly to MLflow!")
```
